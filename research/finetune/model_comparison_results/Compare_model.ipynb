```python

from pathlib import Path
import time
import pandas as pd
from ultralytics import YOLO

# ==================================================
# 1. ĐƯỜNG DẪN MODEL
# Thứ tự khai báo quyết định tên folder kết quả
# ==================================================
MODELS = {
    "YOLO26m": Path(
        r"C:\Users\anhla\Downloads\Web_flask\finetune\train_yolo26m\train-railway-yolo26m.pt"
    ),
    "YOLO11s": Path(
        r"C:\Users\anhla\Downloads\Web_flask\finetune\train_yolo11s\train-railway-yolo11s.pt"
    ),
}

# ==================================================
# 2. ĐƯỜNG DẪN DATASET
# ==================================================
DATASETS = {
    "roboflow_crossing": Path(
        r"C:\Users\anhla\Downloads\dataset_testv3\roboflow_crossing"
    ),
    "rcd_ground": Path(
        r"C:\Users\anhla\Downloads\dataset_testv3\rcd_ground"
    ),
    "rcd_elevated": Path(
        r"C:\Users\anhla\Downloads\dataset_testv3\rcd_elevated"
    ),
}

# ==================================================
# 3. CẤU HÌNH ĐÁNH GIÁ
# ==================================================
IMGSZ = 640
BATCH = 16
CONF = 0.25
IOU = 0.60

# ==================================================
# 4. TẠO FOLDER KẾT QUẢ THEO TÊN MODEL
# Ví dụ: 26m_11s, 26s_11s, 11s_26m
# ==================================================
BASE_OUTPUT_DIR = Path(
    r"C:\Users\anhla\Downloads\Web_flask\finetune\model_comparison_results"
)

def get_model_short_name(model_name):
    return model_name.removeprefix("YOLO")

model_order = list(MODELS.keys())
folder_name = "_".join(
    get_model_short_name(name) for name in model_order
)

OUTPUT_DIR = BASE_OUTPUT_DIR / folder_name
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

print("Thư mục lưu kết quả:", OUTPUT_DIR)

# ==================================================
# 5. TÌM FILE DATA.YAML
# ==================================================
def find_data_yaml(dataset_dir):
    for filename in ("data.yaml", "data.yml"):
        yaml_path = dataset_dir / filename
        if yaml_path.is_file():
            return yaml_path

    candidates = sorted(
        list(dataset_dir.rglob("data.yaml"))
        + list(dataset_dir.rglob("data.yml"))
    )

    if candidates:
        return candidates[0]

    raise FileNotFoundError(
        f"Không tìm thấy data.yaml trong {dataset_dir}"
    )

# ==================================================
# 6. KIỂM TRA ĐƯỜNG DẪN
# ==================================================
for model_name, model_path in MODELS.items():
    if not model_path.is_file():
        raise FileNotFoundError(
            f"Không tìm thấy model {model_name}: {model_path}"
        )

DATA_YAMLS = {}

for dataset_name, dataset_dir in DATASETS.items():
    if not dataset_dir.is_dir():
        raise FileNotFoundError(
            f"Không tìm thấy dataset: {dataset_dir}"
        )

    DATA_YAMLS[dataset_name] = find_data_yaml(dataset_dir)

print("\nCác file YAML được sử dụng:")
for name, path in DATA_YAMLS.items():
    print(f"{name}: {path}")

# ==================================================
# 7. ĐÁNH GIÁ CÁC MODEL TRÊN TỪNG DATASET
# ==================================================
results = []

for model_name, model_path in MODELS.items():

    print("\n" + "=" * 70)
    print(f"ĐANG LOAD MODEL: {model_name}")
    print("=" * 70)

    model = YOLO(str(model_path))

    for dataset_name, yaml_path in DATA_YAMLS.items():

        print(f"\nModel: {model_name}")
        print(f"Dataset: {dataset_name}")
        print("-" * 70)

        run_name = f"{model_name}_{dataset_name}"

        start = time.perf_counter()

        metrics = model.val(
            data=str(yaml_path),
            split="val",
            imgsz=IMGSZ,
            batch=BATCH,
            conf=CONF,
            iou=IOU,
            plots=True,
            project=str(OUTPUT_DIR),
            name=run_name,
            exist_ok=True,
            verbose=True,
        )

        elapsed = time.perf_counter() - start
        box = metrics.box

        result = {
            "Model": model_name,
            "Dataset": dataset_name,
            "mAP50-95": float(box.map),
            "mAP50": float(box.map50),
            "mAP75": float(box.map75),
            "Precision": float(box.mp),
            "Recall": float(box.mr),
            "Validation_time_sec": round(elapsed, 2),
        }

        results.append(result)

        print("\nKẾT QUẢ:")
        print(f"mAP50-95 : {box.map:.4f}")
        print(f"mAP50    : {box.map50:.4f}")
        print(f"mAP75    : {box.map75:.4f}")
        print(f"Precision: {box.mp:.4f}")
        print(f"Recall   : {box.mr:.4f}")
        print(f"Thời gian: {elapsed:.2f} giây")

        # Lưu tiến độ sau mỗi lượt đánh giá
        pd.DataFrame(results).to_csv(
            OUTPUT_DIR / "comparison_progress.csv",
            index=False,
            encoding="utf-8-sig",
        )

# ==================================================
# 8. XUẤT BẢNG TỔNG HỢP
# ==================================================
df = pd.DataFrame(results)

summary_path = OUTPUT_DIR / "comparison_summary.csv"

df.to_csv(
    summary_path,
    index=False,
    encoding="utf-8-sig",
)

print("\n" + "=" * 100)
print("BẢNG SO SÁNH TỔNG HỢP")
print("=" * 100)
print(df.to_string(index=False))

print("\nĐã lưu bảng tổng hợp tại:")
print(summary_path)

# ==================================================
# 9. MODEL TỐT HƠN THEO TỪNG DATASET
# ==================================================
print("\n" + "=" * 100)
print("MODEL TỐT HƠN THEO mAP50-95 TRÊN TỪNG DATASET")
print("=" * 100)

for dataset_name in DATASETS:
    subset = df[df["Dataset"] == dataset_name]

    if subset.empty:
        continue

    best = subset.loc[subset["mAP50-95"].idxmax()]

    print(
        f"{dataset_name}: {best['Model']} "
        f"(mAP50-95 = {best['mAP50-95']:.4f})"
    )

print(f"\nToàn bộ kết quả nằm trong:\n{OUTPUT_DIR}")
```

```python

from pathlib import Path
import shutil
import time
import pandas as pd
import ultralytics
from ultralytics import YOLO

# ============================================================
# 1. CẤU HÌNH ĐƯỜNG DẪN
# ============================================================

PROJECT_DIR = Path(r"C:\Users\anhla\Downloads\Web_flask")

PT_MODEL = (
    PROJECT_DIR / "models" / "yolo26s"
    / "train-railway-yolo26s-all.pt"
)

# Thư mục OpenVINO hiện tại
OPENVINO_SOURCE = (
    PROJECT_DIR / "models" / "yolo26s" / "model_openvino"
)

# Tạo thư mục đúng quy ước Ultralytics
OPENVINO_DIR = (
    PROJECT_DIR / "models" / "yolo26s"
    / "train-railway-yolo26s-all_openvino_model"
)

DATASETS = {
    "roboflow_crossing": Path(
        r"C:\Users\anhla\Downloads\dataset_testv3\roboflow_crossing"
    ),
    "rcd_ground": Path(
        r"C:\Users\anhla\Downloads\dataset_testv3\rcd_ground"
    ),
    "rcd_elevated": Path(
        r"C:\Users\anhla\Downloads\dataset_testv3\rcd_elevated"
    ),
}

IMGSZ = 640
BATCH = 1
WORKERS = 0

# ============================================================
# 2. KIỂM TRA VÀ CHUẨN HÓA THƯ MỤC OPENVINO
# ============================================================

print("Ultralytics version:", ultralytics.__version__)

if not PT_MODEL.is_file():
    raise FileNotFoundError(f"Không tìm thấy PT model:\n{PT_MODEL}")

if not OPENVINO_SOURCE.is_dir():
    raise FileNotFoundError(
        f"Không tìm thấy thư mục OpenVINO:\n{OPENVINO_SOURCE}"
    )

xml_files = list(OPENVINO_SOURCE.glob("*.xml"))

if len(xml_files) != 1:
    raise RuntimeError(
        f"Cần đúng 1 file XML trong {OPENVINO_SOURCE}, "
        f"nhưng tìm thấy {len(xml_files)} file."
    )

xml_source = xml_files[0]
bin_source = xml_source.with_suffix(".bin")

if not bin_source.is_file():
    raise FileNotFoundError(f"Không tìm thấy file BIN:\n{bin_source}")

metadata_source = OPENVINO_SOURCE / "metadata.yaml"

# Sao chép model sang thư mục có tên kết thúc bằng _openvino_model.
# Không xóa hay thay đổi thư mục nguồn.
OPENVINO_DIR.mkdir(parents=True, exist_ok=True)

for source_file in [xml_source, bin_source, metadata_source]:
    if source_file.is_file():
        shutil.copy2(
            source_file,
            OPENVINO_DIR / source_file.name
        )

print("\nThư mục OpenVINO dùng để đánh giá:")
print(OPENVINO_DIR)

print("\nCác file:")
for p in OPENVINO_DIR.iterdir():
    print(" -", p.name)

for name, path in DATASETS.items():
    if not (path / "data.yaml").is_file():
        raise FileNotFoundError(
            f"Không tìm thấy data.yaml của {name}:\n"
            f"{path / 'data.yaml'}"
        )

# ============================================================
# 3. NẠP MODEL
# ============================================================

print("\nĐang nạp model PT...")
model_pt = YOLO(str(PT_MODEL), task="detect")

print("Đang nạp model OpenVINO...")
model_ov = YOLO(str(OPENVINO_DIR), task="detect")

print("Đã nạp thành công hai model.")

# ============================================================
# 4. ĐÁNH GIÁ TRÊN 3 DATASET
# ============================================================

results = []

models = [
    ("YOLO26s PT - CPU", model_pt, "cpu"),
    ("YOLO26s OpenVINO - CPU", model_ov, "cpu"),
]

for dataset_name, dataset_path in DATASETS.items():

    data_yaml = dataset_path / "data.yaml"

    print("\n" + "=" * 80)
    print("DATASET:", dataset_name)
    print("=" * 80)

    for model_name, model, device in models:

        print(f"\nĐang đánh giá {model_name}...")

        start = time.perf_counter()

        metrics = model.val(
            data=str(data_yaml),
            imgsz=IMGSZ,
            batch=BATCH,
            device=device,
            workers=WORKERS,
            plots=False,
            verbose=False,
            split="val",
        )

        elapsed = time.perf_counter() - start

        precision = float(metrics.box.mp)
        recall = float(metrics.box.mr)
        map50 = float(metrics.box.map50)
        map5095 = float(metrics.box.map)

        speed = getattr(metrics, "speed", {}) or {}
        inference_ms = float(
            speed.get("inference", float("nan"))
        )

        fps = (
            1000 / inference_ms
            if 0 < inference_ms < float("inf")
            else float("nan")
        )

        results.append({
            "Dataset": dataset_name,
            "Model": model_name,
            "Precision": precision,
            "Recall": recall,
            "mAP50": map50,
            "mAP50-95": map5095,
            "Inference (ms/image)": inference_ms,
            "FPS suy ra": fps,
            "Validation time (s)": elapsed,
        })

        print(f"Precision : {precision:.4f}")
        print(f"Recall    : {recall:.4f}")
        print(f"mAP50     : {map50:.4f}")
        print(f"mAP50-95  : {map5095:.4f}")
        print(f"Inference : {inference_ms:.3f} ms/image")
        print(f"FPS suy ra: {fps:.2f}")
        print(f"Val time  : {elapsed:.2f} s")

# ============================================================
# 5. HIỂN THỊ BẢNG KẾT QUẢ
# ============================================================

df = pd.DataFrame(results)

print("\n" + "=" * 100)
print("BẢNG SO SÁNH YOLO26s PT VÀ OPENVINO")
print("=" * 100)

display(
    df.style.format({
        "Precision": "{:.4f}",
        "Recall": "{:.4f}",
        "mAP50": "{:.4f}",
        "mAP50-95": "{:.4f}",
        "Inference (ms/image)": "{:.3f}",
        "FPS suy ra": "{:.2f}",
        "Validation time (s)": "{:.2f}",
    })
)

# Bảng so sánh độ trễ theo từng dataset
print("\nSO SÁNH INFERENCE TIME")
display(
    df.pivot(
        index="Dataset",
        columns="Model",
        values="Inference (ms/image)"
    ).round(3)
)

print(
    "\nLưu ý: FPS suy ra = 1000 / thời gian inference (ms/image). "
    "Đây không phải FPS của toàn bộ pipeline."
)
```
